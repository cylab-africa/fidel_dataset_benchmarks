```python
import pandas as pd 
```

```python
df = pd.read_csv('/Users/gabby/Documents/fidel_dataset_benchmarks/src/sample_data/test_labels.csv')
```

```python
df['type'].value_counts()
```

```python
#sample from 30 from each type
sampled_df = df.groupby('type').apply(lambda x: x.sample(n=30, random_state=42)).reset_index(drop=True)
```

```python
# split into train and test sets
train_df = sampled_df.sample(frac=0.8, random_state=42)
test_df = sampled_df.drop(train_df.index)
```

```python
train_df.to_csv('/Users/gabby/Documents/fidel_dataset_benchmarks/src/sample_data/train_labels.csv', index=False)
test_df.to_csv('/Users/gabby/Documents/fidel_dataset_benchmarks/src/sample_data/test_labels.csv', index=False)
```

```python
# I want to provide root of files then use it to copy files in train and test to different root.
#!/usr/bin/env python3
origin_path = '/Users/gabby/Documents/test/'
destination_path = '/Users/gabby/Documents/fidel_dataset_benchmarks/src/sample_data/image_files/'
```

```python
import shutil
for index, row in sampled_df.iterrows():
    file_name = row['image_filename']
    source_file = f"{origin_path}{file_name}"
    destination_file = f"{destination_path}{file_name}"
    shutil.copy(source_file, destination_file)

```
