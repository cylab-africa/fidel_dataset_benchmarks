```python
import pandas as pd
import yaml
```

```python

with open("/home/ubuntu/fidel_dataset_benchmarks/config/main.yaml", "r") as file:
    config = yaml.safe_load(file)
```

```python
for task in ["SYNTHETIC", "HANDWRITTEN", "TYPED"]:
    print("="*30)
    print(task)

    TRAIN_PATH =config[task+"_"+"TRAIN_PATH"]
    VAL_PATH =config[task+"_"+"VAL_PATH"]
    df = pd.read_csv(TRAIN_PATH)


    # Compute length of each line
    df['text_length'] = df['line_text'].astype(str).apply(len)

    # Compute statistics
    stats = {
        'count': df['text_length'].count(),
        'min': df['text_length'].min(),
        'max': df['text_length'].max(),
        'mean': df['text_length'].mean(),
        'std': df['text_length'].std(),
        '25%': df['text_length'].quantile(0.25),
        '50% (median)': df['text_length'].median(),
        '75%': df['text_length'].quantile(0.75),
    }

    # Print nicely
    for k, v in stats.items():
        print(f"{k:<12}: {v:.2f}" if isinstance(v, float) else f"{k:<12}: {v}")
```
