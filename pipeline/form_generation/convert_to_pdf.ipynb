```python

import os
from docx2pdf import convert

def convert_docx_to_pdf(folder_path):
    """
    Converts all .docx files in a given folder to .pdf files.

    Args:
        folder_path (str): The path to the folder containing .docx files.
    """
    try:
        if not os.path.exists(folder_path):
            print(f"Error: Folder '{folder_path}' does not exist.")
            return

        docx_files = [f for f in os.listdir(folder_path) if f.endswith('.docx')]

        if not docx_files:
            print(f"No .docx files found in '{folder_path}'.")
            return

        for docx_file in docx_files:
            docx_path = os.path.join(folder_path, docx_file)
            pdf_file = os.path.splitext(docx_file)[0] + '.pdf' #Change file extension to pdf
            pdf_path = os.path.join('tunga_edits_new', pdf_file)

            try:
                convert(docx_path, pdf_path)
                print(f"Converted '{docx_file}' to '{pdf_file}'")
            except Exception as e:
                print(f"Error converting '{docx_file}': {e}")

        print("Conversion process completed.")

    except Exception as overall_exception:
          print(f"An unexpected error occurred: {overall_exception}")
if __name__ == "__main__":
    # folder_path = input("Enter the folder path containing .docx files: ")  # Get folder path from user
    convert_docx_to_pdf("tunga_edits")
```

```python
import os
import comtypes.client

def convert_docx_to_pdf(input_folder):
    word = comtypes.client.CreateObject("Word.Application")
    word.Visible = False  # Run Word in the background

    for file in os.listdir(input_folder):
        if file.endswith(".docx") and not file.startswith("~$"):  # Ignore temporary files
            doc_path = os.path.join(input_folder, file)
            pdf_path = os.path.join(input_folder, file.replace(".docx", ".pdf"))

            try:
                doc = word.Documents.Open(doc_path)
                doc.SaveAs(pdf_path, FileFormat=17)  # 17 is the PDF format
                doc.Close()
                print(f"Converted: {file} -> {pdf_path}")
            except Exception as e:
                print(f"Failed to convert {file}: {e}")

    word.Quit()
    print("Conversion completed.")

if __name__ == "__main__":
    folder_path = os.path.join(os.getcwd(), "tunga_edits")
    
    if not os.path.exists(folder_path):
        print("Invalid folder path.")
    else:
        convert_docx_to_pdf(folder_path)
```
